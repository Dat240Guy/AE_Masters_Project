CQUAD4 Sympy Calc

```python
import sympy as sp
import numpy as np

xi, eta = sp.symbols('xi eta')
display(xi, eta)
```

```python
N1 = 1/4*(1-xi)*(1-eta)
N2 = 1/4*(1+xi)*(1-eta)
N3 = 1/4*(1+xi)*(1+eta)
N4 = 1/4*(1-xi)*(1+eta)

Coords = np.array([[-1, -1],
                    [1, -1],
                    [1, 1],
                    [-1, 1]])

```

```python
Ns = [N1, N2, N3, N4]
for N in Ns:
    display(sp.simplify(N))
    calc = [N.subs({xi: c[0], eta: c[1]}) for c in Coords]
    print(calc)
    print()
```
