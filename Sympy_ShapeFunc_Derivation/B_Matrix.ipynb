```python
import sympy as sp
from sympy import Derivative as Deriv

x, y, N, i, xi, eta = sp.symbols('x y N i xi eta')

N = sp.Function('N')(x, y)
Nix = Deriv(N, x)
Niy = Deriv(N, y)
display(Nix)
display(Niy)
```

```python
A = sp.Matrix([[1, 0, 0, 0], 
               [0, 0, 0, 1], 
               [0, 1, 1, 0]])

B = sp.Matrix([[Nix, 0],
               [Niy, 0],
               [0, Nix],
               [0, Niy]])

B_matrix = A * B
display(B_matrix)
```

```python
Nnat = sp.Function('N')(xi, eta)
Nixi = Deriv(Nnat, xi)
Nieta = Deriv(Nnat, eta)
display(Nixi)
display(Nieta)
```

```python
A = sp.Matrix([[1, 0, 0, 0], 
               [0, 0, 0, 1], 
               [0, 1, 1, 0]])
display(A)

B = sp.Matrix([[Deriv(xi, x), Deriv(eta,x), 0, 0],
               [Deriv(xi, y), Deriv(eta,y), 0, 0],
               [0, 0, Deriv(xi, x), Deriv(eta,x)],
               [0, 0, Deriv(xi, y), Deriv(eta,y)]])
display(B)

C = sp.Matrix([[Nixi, 0],
                [Nieta, 0],
                [0, Nixi],
                [0, Nieta]])
display(C)
```

```python
display(sp.MatMul(B, C))
```

```python
display(sp.MatMul(B, C).simplify())
```

```python
display(A * B * C)
```
